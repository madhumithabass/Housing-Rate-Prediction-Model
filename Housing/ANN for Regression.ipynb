## Import Libraries

```python
import keras
import numpy as np
```

## Data Collection and Understanding

```python
(X_train,y_train),(X_test,y_test) =  keras.datasets.boston_housing.load_data(path="boston_housing.npz", test_split=0.2, seed=113)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

```python
assert X_train.shape == (404, 13)
assert y_train.shape == (404,)
assert X_test.shape == (102, 13)
assert y_test.shape == (102,)
```

## Data Preprocessing

```python
from sklearn.preprocessing import StandardScaler

Std = StandardScaler()

X_train = Std.fit_transform(X_train)
X_test = Std.fit_transform(X_test)
```

```python
X_train.mean(),X_train.std()
```

```python
X_test.mean(),X_test.std()
```

## Model Building, training,testing and Evaluation

```python
from keras.models import Sequential
from keras.layers import Dense

model = Sequential()
model.add(Dense(units = 64,activation="relu", input_shape=(13,))) 
model.add(Dense(units = 32,activation="relu"))
model.add(Dense(units = 16,activation="relu"))

model.add(Dense(1,)) ## No activation at the output layer
```

```python
model.compile(optimizer='adam',loss="mse",metrics=["mae"])
```

```python
model.fit(X_train,y_train,batch_size=16,epochs=100,validation_split=0.2,verbose=1)
```

```python
y_pred = model.predict(X_test)
```

```python
model.save("ANN_regression.h5")
```

```python
# Example new data (13 features)
new_data = np.array([[0.02, 18, 2.5, 0, 0.45, 6.5, 65, 4.5, 1, 300, 15, 390, 5]])

# Apply same scaling
new_data = Std.transform(new_data)

# Predict
prediction = model.predict(new_data)

print("Predicted House Price:", prediction)
```
